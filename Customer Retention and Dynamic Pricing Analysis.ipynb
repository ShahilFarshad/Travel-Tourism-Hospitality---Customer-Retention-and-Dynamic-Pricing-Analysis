```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_csv("Uncleaned hotel_bookings.csv")

# Preview data
df.head()
```

```python
# Basic Info Check
df.info()
```

```python
df.describe()
```

```python
# Check missing values:
df.isnull().sum()
```

```python
#Fixing datatypes:
#1. Children column stored as float convert it to int

df['children'] = df['children'].fillna(0).astype(int)
```

```python
# replacing null with unknown in country column
df['country'] = df['country'].fillna('Unknown')
```

```python
# coverting data types of columns and replacing missing with 0
df['agent'] = df['agent'].fillna(0).astype(int)
df['company'] = df['company'].fillna(0).astype(int)
```

```python
# convert reservation Status date column into date format
df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'])
```

```python
# remove duplicates
df = df.drop_duplicates()
```

```python
# Remove Invalid Records
# Remove bookings with zero guests
df = df[(df['adults'] + df['children'] + df['babies']) > 0]
```

```python
# Step 7: Handle Outliers
# Remove unrealistic ADR values
df = df[df['adr'] < 5000]
```

```python
#Step 8: Feature Engineering
# Total Guests
df['total_guests'] = df['adults'] + df['children'] + df['babies']
```

```python
# Total Stay
df['total_stay'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
```

```python
# Convert monthname to month_number 
df['arrival_date_month'] = pd.to_datetime(df['arrival_date_month'], format='%B').dt.month
```

```python
# Convert Categorical for ML
df = pd.get_dummies(df, columns=['hotel', 'deposit_type', 'customer_type'], drop_first=True)
```

```python
# Final Check
df.info()
df.head()
df.isnull().sum()
```

```python
# Save Cleaned dataset
df.to_csv("cleaned_hotel_bookings.csv", index=False)
```
