```
df = pd.read_csv("cleaned_hotel_bookings.csv")
df.head()
```

```
plt.figure(figsize=(6,4))

sns.countplot(x='is_canceled', data=df)

plt.title("Booking Cancellation Count")
plt.xlabel("Cancellation Status")
plt.ylabel("Number of Bookings")

plt.show()
```

```
plt.figure(figsize=(6,4))

sns.countplot(x='hotel_Resort Hotel', data=df)

plt.title("Hotel Type Distribution")

plt.xlabel("Resort Hotel (0 = City Hotel, 1 = Resort Hotel)")

plt.ylabel("Count")

plt.show()
```

```
plt.figure(figsize=(12,6))

sns.countplot(x='arrival_date_month', data=df)

plt.xticks(rotation=45)

plt.title("Monthly Booking Trend")

plt.xlabel("Month")

plt.ylabel("Number of Bookings")

plt.show()
```

```
plt.figure(figsize=(10,5))

sns.histplot(df['adr'], bins=50)

plt.title("ADR Distribution")

plt.xlabel("Average Daily Rate")

plt.ylabel("Frequency")

plt.show()
```

```
plt.figure(figsize=(8,5))

sns.boxplot(x='is_canceled', y='adr', data=df)

plt.title("ADR vs Cancellation")

plt.xlabel("Cancellation Status")

plt.ylabel("ADR")

plt.show()
```

```
plt.figure(figsize=(10,5))

sns.histplot(df['lead_time'], bins=50)

plt.title("Lead Time Distribution")

plt.xlabel("Lead Time")

plt.ylabel("Frequency")

plt.show()
```

```
plt.figure(figsize=(10,5))

sns.histplot(df['total_stay'], bins=30)

plt.title("Total Stay Distribution")

plt.xlabel("Total Stay Nights")

plt.ylabel("Frequency")

plt.show()
```

```
plt.figure(figsize=(10,5))

sns.histplot(df['total_guests'], bins=20)

plt.title("Total Guests Distribution")

plt.xlabel("Number of Guests")

plt.ylabel("Frequency")

plt.show()
```

```
plt.figure(figsize=(8,5))

sns.countplot(x='reservation_status', data=df)

plt.title("Reservation Status Analysis")

plt.xticks(rotation=20)

plt.show()
```

```
plt.figure(figsize=(6,4))

sns.countplot(x='is_repeated_guest', data=df)

plt.title("Repeated Guest Analysis")

plt.xlabel("Repeated Guest")

plt.ylabel("Count")

plt.show()
```

```
plt.figure(figsize=(8,5))

sns.histplot(df['total_of_special_requests'], bins=10)

plt.title("Special Requests Distribution")

plt.xlabel("Number of Special Requests")

plt.ylabel("Frequency")

plt.show()
```

```
plt.figure(figsize=(16,12))

sns.heatmap(df.corr(numeric_only=True),
            cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()
```

```
top_country = df['country'].value_counts().head(10)

plt.figure(figsize=(10,5))

top_country.plot(kind='bar')

plt.title("Top 10 Countries by Bookings")

plt.xlabel("Country")

plt.ylabel("Bookings")

plt.show()
```

```
plt.figure(figsize=(8,5))

sns.histplot(df['booking_changes'], bins=20)

plt.title("Booking Changes Distribution")

plt.xlabel("Booking Changes")

plt.ylabel("Frequency")

plt.show()
```

```
plt.figure(figsize=(6,4))

sns.countplot(x='required_car_parking_spaces', data=df)

plt.title("Required Car Parking Spaces")

plt.xlabel("Parking Spaces")

plt.ylabel("Count")

plt.show()
```

Conclusion

- Most hotel bookings were not canceled.
- Seasonal booking patterns were identified across months.
- ADR influenced cancellation behavior.
- Most customers stayed for short durations.
- Repeat guests were lower compared to new guests.
- Business insights from EDA can help hotels improve pricing and retention strategies.
